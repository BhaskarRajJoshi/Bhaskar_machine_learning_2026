```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('dark_background')
plt.rcParams.update({
    'axes.facecolor': '#111111',
    'figure.facecolor': '#111111',
    'savefig.facecolor': '#111111',
    'axes.edgecolor': 'white',
    'axes.labelcolor': 'white',
    'xtick.color': 'white',
    'ytick.color': 'white',
    'text.color': 'white',
    'grid.color': 'gray',
    'legend.facecolor': '#222222',
    'legend.edgecolor': 'white'
})
# Line plot
x = np.linspace(0, 10, 50)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(12, 8), facecolor='#111111')

plt.subplot(2, 2, 1)
plt.plot(x, y1, label='sin(x)', color='blue')
plt.plot(x, y2, label='cos(x)', color='red')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Line Plot: sin(x) vs cos(x)')
plt.legend()
plt.grid(True)

# Scatter plot
plt.subplot(2, 2, 2)
x_scatter = np.random.rand(50)
y_scatter = np.random.rand(50)
colors = np.random.rand(50)
plt.scatter(x_scatter, y_scatter, c=colors, cmap='viridis', s=100)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot')
plt.colorbar()

# Histogram
plt.subplot(2, 2, 3)
data = np.random.randn(1000)
plt.hist(data, bins=30, color='green', alpha=0.7, edgecolor='white')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.grid(True, alpha=0.3)

# Bar chart
plt.subplot(2, 2, 4)
categories = ['Math', 'Python', 'AI', 'ML', 'DS']
scores = [85, 90, 88, 92, 87]
plt.bar(categories, scores, color='purple', alpha=0.7, edgecolor='white')
plt.xlabel('Subject')
plt.ylabel('Score')
plt.title('Bar Chart - Subject Scores')
plt.ylim(0, 100)

plt.tight_layout()
plt.show()

print("Visualization complete!")
```

```python
# Creating a DataFrame
data = {
    'Name': ['Bhaskar', 'Raj', 'Joshi', 'Alice', 'Bob'],
    'Age': [20, 22, 21, 23, 19],
    'Score': [85.5, 90.0, 77.5, 88.0, 92.5],
    'Grade': ['B', 'A', 'C', 'A', 'A']
}

df = pd.DataFrame(data)
print("DataFrame:")
print(df)

# Basic DataFrame operations
print(f"\nDataFrame shape: {df.shape}")
print(f"DataFrame info:")
print(df.info())

print(f"\nDataFrame description:")
print(df.describe())

# Selecting columns
print(f"\nName column:\n{df['Name']}")
print(f"\nAge and Score columns:\n{df[['Age', 'Score']]}")

# Filtering data
print(f"\nStudents with score > 85:")
print(df[df['Score'] > 85])

print(f"\nStudents with Grade A:")
print(df[df['Grade'] == 'A'])

# Adding new column
df['Pass'] = df['Score'] >= 60
print(f"\nDataFrame with Pass column:")
print(df)

# Handling missing data (example)
df_with_nan = df.copy()
df_with_nan.loc[0, 'Score'] = np.nan
print(f"\nDataFrame with missing data:")
print(df_with_nan)
print(f"Rows with missing data:\n{df_with_nan[df_with_nan['Score'].isna()]}")
print(f"\nFilled missing data (forward fill):")
print(df_with_nan.fillna(method='ffill'))
```

```python
# Try-Except block for handling exceptions
try:
    num = int("not a number")
except ValueError:
    print("Error: Could not convert string to integer!")
except Exception as e:
    print(f"An unexpected error occurred: {e}")

# Handling multiple exception types
try:
    numbers = [1, 2, 3]
    print(numbers[0])  # This works
    print(numbers[10])  # This raises IndexError
except IndexError:
    print("Error: List index out of range!")
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

# Try-Except-Finally
try:
    file = open("nonexistent.txt", "r")
except FileNotFoundError:
    print("Error: File not found!")
finally:
    print("This always executes, whether an error occurred or not.")

# Raising custom exceptions
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")
    elif age < 18:
        return "Minor"
    else:
        return "Adult"

try:
    print(check_age(25))  # This works
    print(check_age(-5))  # This raises ValueError
except ValueError as e:
    print(f"Error: {e}")
```

## 5. Control Flow: Loops and Conditionals

Practice if-else statements, for loops, while loops, and nested control structures.

```python
# Define a function with parameters and return value
def calculate_hypotenuse(a, b):
    """Calculate hypotenuse using Pythagorean theorem"""
    return sqrt(a**2 + b**2)

result = calculate_hypotenuse(3, 4)
print(f"Hypotenuse of 3 and 4: {result}")

# Function with default parameters
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Bhaskar"))
print(greet("Bhaskar", "Hi"))

# Lambda expressions (anonymous functions)
square = lambda x: x**2
cube = lambda x: x**3

print(f"\nLambda functions:")
print(f"Square of 5: {square(5)}")
print(f"Cube of 5: {cube(5)}")

# Using lambda with map and filter
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
evens = list(filter(lambda x: x % 2 == 0, numbers))

print(f"Squared numbers: {squared}")
print(f"Even numbers: {evens}")
```
